# Sequential Workflow: Simple LLM Q&A

`START -> llm_qa -> END`

```python
from langgraph.graph import StateGraph, START, END
from langchain_groq import ChatGroq
from typing import TypedDict
from dotenv import load_dotenv

load_dotenv()

model = ChatGroq(model='openai/gpt-oss-20b')
```

```python
class LLMState(TypedDict):
    question: str
    answer: str
```

```python
def llm_qa(state: LLMState) -> LLMState:
    prompt = f"Answer the following question in 2-3 lines.\n{state['question']}"
    state['answer'] = model.invoke(prompt).content
    return state
```

```python
graph = StateGraph(LLMState)

graph.add_node('llm_qa', llm_qa)

graph.add_edge(START, 'llm_qa')
graph.add_edge('llm_qa', END)

workflow = graph.compile()
```

```python
result = workflow.invoke({'question': 'How far is the Moon from the Earth?'})
print(result['answer'])
```
